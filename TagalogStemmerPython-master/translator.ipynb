```python
import pandas as pd
from deep_translator import GoogleTranslator
from tqdm import tqdm  # Import tqdm for the progress bar

# Set the local file path of the original CSV
csv_path = 'C:/Users/danjo/Downloads/preprocessed_tagalog_data (3).csv'  # Update this path
df = pd.read_csv(csv_path)

# Function to translate text
def translate_text(text):
    try:
        translator = GoogleTranslator(source='tl', target='en')  # Translate Tagalog to English
        return translator.translate(text)
    except Exception as e:
        return f"Error: {e}"

# Find the index of the 'emotion' column
emotion_index = df.columns.get_loc("emotion")

# Apply translation with a progress bar
tqdm.pandas(desc="Translating text")
df["translated_processed_text"] = df["processed_text"].progress_apply(translate_text)

# Insert the new column before 'emotion'
df.insert(emotion_index, "translated_processed_text", df.pop("translated_processed_text"))

# Save the new CSV file
output_path = 'C:/Users/danjo/Downloads/Translated-preprocessed_tagalog_data (3).csv'  # Update this path
df.to_csv(output_path, index=False)

print(f"✅ Translation complete. File saved as: {output_path}")
```
